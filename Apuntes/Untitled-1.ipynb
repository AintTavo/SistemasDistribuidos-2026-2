## CORBA

Common Object Request Broker Arquitecture

Es una aquitectura que permite diferentes programas se comuniquen entre si, aunque esten hechos con distintos lenguajes de programación

Fue la base de los sistemas distributivos actuales, ocupando puertos dinamicos y con el auge de la seguridad del internet con su mayor uso, empezo a ser bloqueado por firewall. Al ocupar TCP/IP y su dependencia fuerte a este y como este dividia las tramas causaba problemas en el envio de este. Además como tenia muchos modelos basados unicamente en transmición la recepción se vuelve dificil. 

Al ver que se podian mandar datos unicamente por JSON y XML, ver que era mas facil mandar los metadatos del objeto que los objetos completos

## ¿Para que sirve?

Conectar sisetmas distribuidos, para permitir interoperabilidad, facilitar la comunicación entre objetos remotos.

### Componenetes principales de CORBA

- ORB Object REquest Broker: Encargado de manejar las peticiones.
- IDL Interface Definition Language : Encargado de definir las interfaces de cada objeto.
- Registro de interfaces Interface REpository : Interfaces dentro del esquema (opcional).
- Cliente y servidor

### Secuencua

Cliente -> ORB(Struct) -> ORB (localiza al objeto que puede realizar la operación) -> Objeto (Realiza la operación) -> ORB (Envia el resultado al cliente)  -> Servidor

## COM / DCOM

PAso de COM a DCOM

COM permitia la comunicacion mientras sea en el mismo equipo.
DCOM funciona dando la metodo de ocupar proxys y stops como intercambio entre cliente y servidores.
MEcanismos de marshalling, este empaqueta los datos en para pasarlos por la red.
Ciclo de vida pinging. Manda señales de que el servidor y usuario sigue con vida cada cierto tiempo
