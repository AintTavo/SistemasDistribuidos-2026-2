# Evolución de las Arquitecutras Distribuidas: Un Análisis Técnico de Servicios Web, Aplicaciones Web Progresivas y Microservicios

La ingenieria de software contemporánea ha experimentado una transición paradigmática desde las estructuras monolíticas centralizadas hacia ecosistemas distribuidos, modulares y altamente resilientes. Esta trasnformación no es meramente estéticas ni una tendencia pasajera, sino una respuesta técnica a la necesidad de escalabilidad masiva, disponibilidad global y agilidad en el ciclo de vida del desarrollo. Para comprender el panorama actual, es imperativo desglosar tres oulares fundamentades que a menudo confunden pero que cumplen roles distintos en la pila tecnoógica: los Servicios Web(API), las Aplciaciones Web Progresivas (PWA) y la arquitectura de Microservicios. Cada uno de estos conceptos representa una capa de abstracción diferente; mientras que las API actúan como el tejido conectivo o el contrato de comunicación entre sistemas, las PWA redefinen la experiencia del usuario final en el frontend, y los microservicios proponen una reorganización estructural y organizaciones del backend.

La relevancia de estos análisis radica en la capacidad de discerner cuándo y cómo implementar cada tecnología para optimizar recursos y garantizar la sostenibilidad del softwaare. Un profesional con conocimientos de programación pero que no ha interactuado directamente con estas arquitecturas debe comprender que la trasnsición hacia lo distribuido conlleva una complejidad inherente que debe ser gestionada mediante patrones de diseño rigurosos y una infraestructura robista.

# Servicios WEB y las API

En el contexto de la computación moderna, un servicio web es un mecanismo de software que permite la interacción entre dos máquinas a travpes de una red, utiliazndo un formato de datos estandarizado y protocolos de comunicación conocidos. La interfaz de Programación de Aplicaciones (API), por su parte, es el conjunto dereglas, definiciones y protocolos que permiten que una aplicación se comunique con otra. En escencia, toda API de servicio web es una interfaz, pero no todas las API son servicios web; sin embargo, en la industria actual, ambos términos se utilizan casi indistintamente para referirse a la capa de exposición de datos y funciones de un servidor:

## Filosofías de Diseño: SOAP, REST y GrapgQL

El desarrollo de las APi ha evolucionado a través de tres grande sfilosofías de diseño, cada una con sus propias restricciones y casos de uso óptimos.

### SOAP

SOAP (Simple Object Access Protocol) respresentea el enfoque más tradicional y estructurado. Basado exclusivamente en XML para intercambio de mensajes, SOAP se rige por un contrato estricto denominado WSDL(Web Services Description Language). Este archivo WSDL actúa como una "fuente de verdad" única que describe cada operación, parámetro y tipo de dato que el servicio soporta. La rigidez de SOAP es su mayor fortaleza en entornos empresariales críticos, como la banca o la salud, donde la seguridad (a traés de estándates como WS-Security) y la intergirad de las transacciones son innegociables. No obstante, su verbosidad y la pesadez de los mensajes XML suele penalizar el rendimiento enr edes con ancho de banda limitado.

### REST

REST (Represntational State Transfer) surgión como una alternativa más ligera y flexible, aprovechando los principios fundamentales del protocolo HTTP. A diferencia de SOAP, que se centra en peraciones o funciones, REST se basa en recursos. Cada entidad (un usaurio, unpedido, un producto) es identificada con una URI única y manipulada mediante métodos HTTP estándar: GET para recuperación, POST para creación, PUT o PATCH para actualización y DELETE para eliminación. La flexibilidad de REST permite el uso de múltiples formatos, aunque JSON se ha consolidado como el estándar de fato debido a su ligereza y facildiad de procesamiento por parte de los motores de JavaScript en los navegadores.

### GraphQL

GraphQL representa el estado del arte en la consulta de datos, diseñado para resolver los prolemas de "over-fetching" (recibir mś datos de los necesarios) y "under-fetching" (tener qyue hacer múltiples peticiones para obtener datos relacionados). En lugar de múltiples puntos de enlace (endpoints) fijos, GraphQL ofrece un único punto de entrada donde el cliente envía una consulta que define exactamente qué campos y qué relaciones desea recibir. El servidore responde ocn un objeto JSON que espeja la estructura de la consulta, optimizando drásticamente la transferencia de datos.

## Funcionamiento de servicios web

El funcionamiento de un servicio web se basa en un cliclo de solicitud-respuesta sin estado (stateless). Esto significa que cada interacción entre el cliente y el servidor debe contener toda la información necesaria para ser procesada, sin depender de un contexto guardado en el servidor de solicitudes anteriores.

| Fase | Acción Técnica | Componentes Involucrados |
| --- | --- | --- | 
| Solicitud | El cliente contruye un paquete HTTP con Verbo, URI, Headers (Auth, Contetn-Type) y Body (si aplica) | Cliente (App, Browser), HTTP Stack |
| Transporte | El mensaje viaja cifrado por TLS a través de internet hasta el endpoint del servidor | DNS, TLS/SSL, Proxies |
| Procesamiento | El servidor valdia la autenticación, parsea el cuerpo, ejecuta la lógica y consulta la base de datos. | Servidor Web, Lógica de Negocio, BD
| Respuesta | El servidor genera un código de estado (2xx, 4xx, 5xx) y devuelve la caga útil (Payload) en JSON o XML | Api Handler, Serializador |



## Requisitos para la implementación de una API

Para que un desarrollador implemente un servicio web robusto, es neceario contar con un entorno de ejecución capaz de manejar concurrencia y conexiones de red.

1. **Entorno de Ejecución y Framework:** Dependiendo del lenguaje, se utilizan herramientas como Nod.js con Express, Java con Spring Boot, o Python con FastAPI. Estos proporcionan las abstracciones para manejar rutas HTTP.

2. **Seguridad y Autenticación:** Se requiere la implementación de esquemas como JWT (Json Web Tokens) o OAuth2.0 para asegurar que solo clientes autorizados accedan a los recursos.

3. **Gestión de Datos:** Una base de datos (SQL o NoSQL) que persista la información y una capa de accesso a datos (ORM o consutlas directas).

4. **Servidor de Aplciaciones y Certificados SSL:** El uso de HTTPS es obligatorio en la web moderna para proteger los datos en tránsito.

## Proceso de Implementación y Pseudocódigo

La implementación de una API comienza con el diseño del modelo de recursos. Supongamos un sistema de gestión de productos. El primer paso es definir las rutas (endpoints) y la lógica asociada a cada verbo HTTP.

### Pasos Lógicos

1. Inicializar el servidor y configurar el middleware para interpretar JSON.

2. Definir la estrutura del objeto Producto.

3. Crear manejadores de rutas (Handlers) para cada endpoint.

4. Intergrar la validación de datos para asegurar la integridad de la entrada.

5. Configurar el puerto de escucha y las excepciones globales.

### Pseudocódigo de una API REST (Estilo Express)

```javascript
// Carga del motor del servidor
importar modulo_servidor;
instancia app = modulo_servidor();

// Middleware para transformar cuerpos de peticiones en objetos manipulables
app.usar(interprete_json());

// Simulacion de almacenamiento persistente
variable productos =;

// OPERACION: Obtener todos los recursos (READ)
app.get("/api/v1/productos", funcion(solicitud, respuesta) {
    // Retornar lista completa con codigo de exito 200
    enviar_respuesta(200, productos);
});

// OPERACION: Crear un recurso (CREATE)
app.post("/api/v1/productos", funcion(solicitud, respuesta) {
    variable nuevo_dato = solicitud.cuerpo;
    
    // Validacion logica simple
    si (nuevo_dato.nombre es_nulo O nuevo_dato.stock < 0) {
        enviar_respuesta(400, { mensaje: "Datos de producto invalidos" });
    } sino {
        nuevo_dato.id = generar_nuevo_id();
        productos.agregar(nuevo_dato);
        enviar_respuesta(201, nuevo_dato); // 201 significa 'Creado'
    }
});

// Inicio de recepcion de trafico
app.escuchar_en_puerto(8080, funcion() {
    imprimir("Servicio API operando en puerto 8080");
});
```




# PWA

Una Aplicación Web Progresiva no es un lenguaje de programación nuevo, sino una metodología de desarrollo que utiliza tecnologías web estándar (HTML, CSS, JavaScript) para proporcionar una experiencia que tradicionalmente solo estaba disponible en aplicaciones nativas. La filosofía detrás de una PWA es que la web debe ser capaz de todo: funcionar sin conexión, enviar notificaciones y ser instalable, manteniendo al mismo tiempo la facilidad de actualización y el alcance masivo de un sitio web

## El Motor Tecnológico: Manifest y Service Worker

La transformación de un sito web convencional en una PWA depende de dos pilares técnicos que actúan como puentes entre el navegador y el sistema operatvio.

### Manifest

El **Manifiesto de la Aplicación** (webmanifest.json) es un archivo de metadatos que infrma al navegador que el sito web califica para ser instalado como una aplicación. En este archivo se definen las propiedades visuales: el nombre corto para el escritorio, los iconos en diferentes resoluciones, el colo de fondo de la pantalla de carga (splash screen) y, lo más importante, el modo de visualización. Al configurar el modo "display":"standalone", la aplicación se lanza sin la barra de direcciones del navegador, proporcionando una inmersión total similar a una app nativa.

### Service Worker

El **Service Worker** (sw.js) es quizás el avance más significatio en la web en la última década. Es un script de JavaScript que el navegador ejecuta en un hilo separado de la interfaz de usuario. No tiene acceso directo al DOM (Documento Object Model), pero tiene la capcidad de interceptar todas las solicitudes de red que salen de la aplicación. Al actuar como un proxy programable, el Service Worker puede decidir si una imagen o un script debe descargarse de internet o si debe servirse instantáneamoente desde un caché local.

## Funcionamiento: Ciclo de Vida y Estrategias de Almacenamiento

El Service Worker opera bajo un ciclo de vida estricto gestionado por el navegadfor para garantizar que las actualizaciones de la aplicación no crrompan el estado del usuario.

1. **Registro:** El hilo principal de la aplcación detecnta si el navegador soporta Service Workers y apunta al archivo correspondiente.

2. **Instalación:** Se dispara el evento install. Aquí suelde definir una lsita de archivos críticos (App Shell) para guardar en la caché de forma inmediata.

3. **Activación:** Una vez instalado, el Service Worker toma control de la aplciación. Es el momento de purgar versiones antiguas de la caché para liberar espacio.

Las estrategias de caché son fundamentalas para la experiencia del usuario:

| Estrategia | Descripción | Caso de Uso |
| --- | --- | --- |
| **Cache First** | Busca el recurso en la caché. Si no existe, va a la red. | Activos estáticos (CSS, Imágenes, JS) |
| **Network First** | Intenta obtener el dato fresco de la red. Si falla (offline), usa la caché. | Datos dinámicos (noticias, perfil de usuario).
| **Stale-While-Revalidate** | sirve de inmediato lo que hay en la caché, pero descarga la versión nueva en segundo plano para la próxima vez. | Balance ideal entre velocidad y frescura. |





## Requisitos y proceso de implementación de una PWA

La implementación exitosa de una PWA requiere cumplir con estándares de calidad que los navegadores verifican antes de permitir la instalación.

- **Protocolos HTTPS:** Por razones de seguridad, los Service Workers solo funcionan en conexiones cifradas (excepto en localhost para desarrollo), ya que tienne el poder de modificar las respuestas de red.

- **Diseño Adaptable (Responsive):** La aplicación debe funcionar perfectamente en dispositivosmóviles, tablets y escritorios.
- **Conectividad Offline:** Al menos una página de respaldo debe estar disponible cuando no hay red.

El proceso consiste en desarrollar la web tradicionalmente, añadir el archivo manifiesto, programar el script del Service Worker con lógica de caché y, finalmente, registrar dicho script en el archivo JavaScript principal de la aplicación.

### Pesudocódigo de una PWA (Implementación de Service Worker)

Para un programador, la lógica de un PWA se divide entre lo que ocurre en el cliente visible y lo que ocurre en el hilo del trabajador.

**Registro en el cliente (main.js)**
```javascript
si ('serviceWorker' en navegador) {
    ventana.al_cargar(() => {
        navegador.serviceWorker.registrar('/trabajador-red.js')
           .entonces(reg => imprimir("PWA lista: Trabajador registrado"))
           .error(err => imprimir("Error en registro PWA", err));
    });
}
```

**Lógica del Trabajador de REd (trabajador-red.js)**
```javascript
const NOMBRE_VERSION = "cache-sistema-v1";
const ARCHIVOS_CRITICOS = [
    "/index.html",
    "/css/estilos_globales.css",
    "/js/logica_principal.js",
    "/img/icono_app.png"
];

// FASE 1: INSTALACION - Guardar el esqueleto de la app
escuchar_evento('install', (evento) => {
    evento.esperar_hasta(
        abrir_almacen_cache(NOMBRE_VERSION).entonces(cache => {
            imprimir("PWA: Almacenando archivos base para uso offline");
            retornar cache.agregar_lista(ARCHIVOS_CRITICOS);
        })
    );
});

// FASE 2: INTERCEPCION - Controlar el trafico de red
escuchar_evento('fetch', (evento) => {
    evento.responder_con(
        buscar_en_cache(evento.solicitud).entonces(respuesta => {
            // Si el archivo esta en el dispositivo, lo entregamos sin red
            si (respuesta_existe(respuesta)) {
                retornar respuesta;
            }
            // Si no, lo descargamos normalmente
            retornar descargar_de_internet(evento.solicitud);
        })
    );
});
```

# Microservicios

La arquitectura de microservicios es un estilo de siseño donde una aplciación se construye como una suite de servicios pequeños, autónomos e independientes, cada uno de los cauels se ejecuta en su propio proceso y se comunica con los demás a ttravés de protocolos ligeros. Mientras que en una arquitectura monolítica todas las funciones (pagos, inventario, usuarios) comparten la misma base de código y la misma base de datos, en los microservicioes cada función empresarial es unsistema completo por sí mismo.

## Principios Fundamentales y Estructura Interna

La modularidad de los microservicios no es solo a nivel de código, sino también a nivel de datos y organización.

1. **Desacoplamiento Total:** Cada microservicio puede desarrollarse, desplegarse y escalarse sin afectar a los demás. Esto permite utiliazr diferentes lenguajes de programación según la necesidad (Políglota).

2. **Base de Datos Independiente:** El principio de "Database per Service" dicta que un microservicio no debe acceder directamente a la base de datos de otro. La comuncación de datos siemrpe debe ocurrir a través de API.

3. **Responsabildiad Única:** Un servicio debe hacer una sola cosa y hacerla bien (Single Responsability Principle).

Un microservicio típico se estructura internamente en tres capas bien definidas para asegurar su mantenibilidad:

| Capa | Función Técnica | Implementación Común |
| --- | --- | --- |
| Interfaz(API) | Punto de entrada para solicitudes externas e internas. | REST Controllers, gRPS Endpoints |
| Lógica de Negocio | Procesa las reglas y cálculos específicos del dominio. | Clases de Servicio, Casos de Uso. |
| Persistencia(Datos) | Almacenamiento de la información necesaria para el servicio. | DB SQL (PostgreSQL) o NoSQL (Redis/Mongo) |

## Comunicación entre Microservicios: Síncorna vs Asíncorna

Debido a que el sistema está distribuido, la forma en que los servicios hablan entre sí es crítica para el rendimiento y la tolerancia a fallos.

- **Comunicación Síncorna:** Un servicio llama a la API de otro y espera la respuesta (bloqueante). Se utiliza comúnmente con protocolos como HTTP/REST o GRPC. Es fácil entender perpo puede causar "caídas en cascada" si un servicio lento bloquea a toda la cadena de llamadas.

- **Comunicación Asíncorna:** Un servicio envía un mensaje a un intermediario (Message Broker como RabbitMQ o Kafka) y continúa con su ejecución sin esperar respuesta. otros servicios interesados se suscriben a ese canal de mensajes y procesan la información a su propio ritmo. Este modelo es mucho más resiliente y escalable.

## Infraestructura y Orquestación

Mantener decenas de microservicios ejecutándose simultáneamente require de herramientas de automatización avanzadas.

- **Contenedres (Docker):** Cada microservicio se empaqueta en una "caja" virutal que contiene todo lo necesario para correr (código, runtime, librerías). Esto garantiza que el servicio funcione igual en cualquier servidor.

- **Orquestadores (Kubernetes):** Es el sistema que gestiona el despliege masivo de contenedores, encargándose de reiniciarlos si fallan, balanear el tráfico y escalar el número de instancias seǵun la demanda.

- **API Gateway:** Un servidor que actúua como puerta única para el mundo exterior. Se encarga de recibir todas las peticiones de las aplicaciones clientes y enrutarlas al microservicio correcto, manejando además la seguridad y el límite de tráfico (Rate Limiting).



### Pesudocódigo de Microservicios (Modelo Asincorno con Broker)

Ete ejemplo describe la interacción entre un Microservicio de Órdenes y un Microservicio de Inventario a través de una cola de mensajes.

**Microservicio de Órdenes (Productor):**
```javascript
clase OrdenController {
    metodo confirmar_compra(detalle_orden) {
        // 1. Guardar en base de datos local
        orden_id = db_ordenes.guardar(detalle_orden);
        
        // 2. Preparar mensaje para el sistema distribuido
        mensaje_evento = {
            id: orden_id,
            items: detalle_orden.productos,
            estado: "PENDIENTE_INVENTARIO"
        };
        
        // 3. Publicar en el Bus de Mensajes (RabbitMQ/Kafka)
        // No esperamos respuesta, el sistema continua
        bus_mensajes.publicar("cola_validar_stock", mensaje_evento);
        
        retornar { status: "Procesando", id: orden_id };
    }
}
```

***Microservicio de Inventario (Consumidor):**
```javascript
clase InventarioWorker {
    // Este proceso corre en segundo plano escuchando la cola
    escuchador_de_mensajes.al_recibir("cola_validar_stock", (datos) => {
        imprimir("Validando stock para orden: " + datos.id);
        
        variable exito = inventario_db.descontar_items(datos.items);
        
        si (exito) {
            bus_mensajes.publicar("cola_orden_confirmada", { id: datos.id });
        } sino {
            bus_mensajes.publicar("cola_orden_rechazada", { id: datos.id, motivo: "Sin stock" });
        }
    });
}
```

# Comparativa

| Critero | Servicio Web (API) | App/PWA | Microservicios |
| --- | --- | --- | --- |
| **Enfoque principal** | Intercambio de datos. | Experiencia de usuario(UX). | Escalabilidad y mantenibilidad. |
| **Unidad de Despliege** | Servidor de Aplicación. | Navegador del Cliente. | Contenedores(Docker/K8S) |
| **Escalabilidad** | Vertical (más CPU) o Horizontal simple. | No aplica(depende del dispositivo). | Horizontal Selectiva (por servicio). |
| **Persistencia** | Base de datos centralizada. | Local (IndexedDB) + Remota. | Base de datos por servicio. |
| **Curva de Aprendizaje** | Baja (Conceptos HTTP). | Media(Service Workers/Offline). | Alta (Infraestructura/Redes). |
| **Tolerancia a Fallos** | Si falla el API, falla el sistema. | Funciona offline(degradación parcial). | Resiliente; si un servicio cae, el resto sigue vivo. |
| **Costo Inicial** | Bajo. | Medio(Un solo código base). | Muy Alto (Requiere equipo DevOps). |
