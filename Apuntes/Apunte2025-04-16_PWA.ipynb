## Service Workers

Un Service Worker es un script que permite interceptar y controlar como un navegar web maneja las solicitudes de red y el almacenamiento en caché de activos. Teniendo funcionamiento sin dependencia al internet. Teniendo diferencias como capa intermedia de broker, llamada Service Worker.
