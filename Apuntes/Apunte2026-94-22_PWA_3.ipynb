# Progresive

Mantener el accesos por medio de la app, la pagina se puede mandar como aplicación nativa por medio del navegador. Principalmente ocupando sobre los dispositivos moviles, este puede ser ocupado como una aplicación nativa pero principalmnente se hace una llamada sin ui del navegodar la cual se manda y descarga los datos primordiales para su ejecución sin internet. Para esto guarda todo en cache y un semi servidor local, esto hace que sea aparte de portable para distintos dispositivos. 

Esto la hace depediende de la cache del navegador base, esto hace que los recursos del cliente hace que tengan un gssto mayor del lado del cliente, ademas de gastar mas cache del navegador. Esto es solucionable cada vez que el navegador vuelve a consultar datos seria principalmente gestionar de mejor manera esta con cada vez que resaliza esto.

Otro problema son los permisos de la aplicación, si puede tener los permisos, pero esta tiene que tomar una aplicación externa dependiendo del dispostivo para realizar esta tarea. Mientras que una aplicación completamente nativa puede tener acceso a estos recursos dentro de la misma aplicación. Esto añade un pequeño delay por tener que llamar a otra aplicación.

Además otro problema anexo a las PWAs es que estas deben estar completamente bien empaquetadas con todas sus librerias necesarias. Debido a que dependiendo del navegador predeterminado del dispositivo final haga que este no se ejecute correctamente o tenga errores graficos. Por esto es necesario incluir con ella todas las librerias que necesita de forma directa
