# APIS
 
Lo complicado era hacer toda la protocolizacion y hacer que la comunicación fuera estandar entre todas las partes, y centralizar el broker


PAra las apis fue necesario quitar la salidas a la red por parte de un solo servidor coordinador unicamente por medio de peticiónes de POST y GET, o peticiones estandar las cuales solo llaman a un servicios

LAs peticiones pueden ser a tipo funcional o tipo de recurso

- Un servicio funcional se trabajan con funciones que estan esperando peticiones dentro de un sistema.
- Un servicio de recurso permite ejecutar partes de codigo o almacenar datos.

Las APIs surgen de ocupar protocolos estándar entre todos los sistemas, debido a que antes todos los protocolos eran propietarios. Con la llegada de TCP / IP al hacerlo estándar para toda la población ya no fueron necesarias las tecnológias como los objetos distribuidos dado que estos eran los intermediaros entre el internet general a un sistema propietario. A solo necesitar un mensaje que todos entiendan, algo que tambien necesitaba los objetos distribuidos, para esto se ocuparon tecnologias estándar de igual manera mandando la información para esto surgieron tecnologías como XML y JSON.
